```python
def lesser_of_two_evens(a,b):
    if a%2 == 0 and b%2 == 0: # Se usa modulo de primero para saber si alguno es primo o no
        return min(a,b)
    else: # Si ambos son pares, se busca el menor con min, si no, el mayor con max
        return max(a,b) 


print(lesser_of_two_evens(2,4))
print(lesser_of_two_evens(2,5))

print("\n")

#Galletas

def animal_crackers(text):
    letra1 = text[0]
    text2 = text.split() # Se separa el texto en palabras para comparar la primera letra de cada una
    letra2 = text2[1][0] # Se guarda la 1ra letra de la segunda palabra

    if letra1 == letra2: # Compara si ambas palabras inician con la misma letra
        return True
    else:
        return False

print(animal_crackers('Ciudad Calor'))
print(animal_crackers('Lugar Frio'))

#DADOS
print("\n")

def makes_twenty(n1,n2):
    if n1 == 20 or n2 == 20: # Si alguno de los dos ya es 20 se cumple directamente
        return True
    elif (n1+n2) == 20: # También se cumple si la suma de ambos da 20
        return True
    else:
        return False # si no, retorna falso

print(makes_twenty(20,10))
print(makes_twenty(2,3))
print(makes_twenty(12,8))

#mayusculas

print("\n")

def old_macdonald(name):
    new1 = name.capitalize() # Pone en mayúscula solo la primera letra de todo el nombre

    separate = new1[3:len(new1)]  # Se separa desde la posición 3 en adelante para volver a capitalizar esa parte. Se hace con slicing
    cap = separate.capitalize() # se capitaliza la parte cortada

    final = new1.replace(new1[3:len(new1)], cap) # Reemplaza esa parte del texto por la versión con mayúscula, logrando 2 mayúsculas en el nombre

    return final

print(old_macdonald('macdonald'))

#reversa
print("\n")

def master_yoda(text):
    lista = text.split() # se separa todo para guardarlo en una lista
    n = len(lista) -1 # Se resta 1 porque el índice del último elemento de la lista es len-1
    yoda = ''

    for i in range(n,-1,-1): # Recorre la lista de atrás hacia adelante para invertir el orden de las palabras
        yoda +=lista[i] # Y se guarda una a una cada palabra a la inversa
        yoda += ' ' # se usa para poner un espacio entre cada palabra
    return yoda

print(master_yoda('I am home'))
print(master_yoda('We are ready'))

#un 3 seguido
print("\n")

def has_33(nums):

    for i in range(0, len(nums) - 1):
        if nums[i] == 3 and nums[i] == nums[i+1]:
            return True # se revisa si hay 3 seguidos con i+1

    return False # si no encuentra nunca, sale del for y tiene que retornar falso

print(has_33([1, 3, 3]))
print(has_33([1, 3, 1, 3]))
print(has_33([3, 1, 3]))

#Replicador
print("\n")

def paper_doll(text):
    resultado = ""
    for letra in text: # Recorre cada carácter del texto para triplicarlo
        resultado += letra * 3 # se usa una nueva cadena y se usa el *3, ya que ya vimos que estos operadores funcionan igual para enteros y letras
    return resultado

print(paper_doll('Hello'))
print(paper_doll('Mississippi'))

#BlackJack
print("\n")

def blackjack(a,b,c):
    suma = 22 # Valor por defecto que garantiza 'Bust!' si ninguna condición del if se cumple

    if a+b+c <= 21:
        return a+b+c
    elif a+b+c > 21 & a == 11 or b == 11 or c == 11:
        suma = a+b+c-10 # Si hay un As (11), se puede restar 10 para bajar la suma sin pasarse, ya que el as puede ser 1
    
    if suma >= 21:
        return 'Bust!'
    else:
        return suma


print(blackjack(5,6,7))
print(blackjack(9,9,9))
print(blackjack(9,9,11))
```
