## Actividad: Clase Elipse

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

class Ellipse(object):

    # Constructor
    def __init__(self, width, height, color1='r', color2='b'):
        self.height = height
        self.width = width
        self.color1 = color1 # color1 es el color de relleno del elips
        self.color2 = color2 # El otro color es el exterior del elipse
        self.xy = (0,0) # Se inicializa en 0,0 como base para que aparezca en la mitad del grafico
     
    def setWidth(self, width): # Los set ayudan a cambiar los datos existentes sin instanciar otra vez el elipse
        self.width = width #se usa el self para que se sepa que es el dato de la propia clase

    def setHeight(self, height):
        self.height = height

    def setcolor1(self, color1):
        self.color1 = color1
    
    def setcolor2(self, color2):
        self.color2 = color2

    # Method
    def drawEllipse(self): # Se usa con los datos de la propia clase. se usa self para que se especifique que son con esos
        plt.gca().add_patch(patches.Ellipse(self.xy, width = self.width, height = self.height, facecolor = self.color1, edgecolor = self.color2))
        plt.axis('scaled') # Scaled se usa para que aparezca todo el elipse y la imagen se adapte y no salga solo una parte de este
        plt.show()


ellipse = Ellipse(1, 2, 'red', 'blue')
ellipse.drawEllipse() # Creamos la estancia de elipse con un tamaño pequeño y llamamos el método
# que grafica el propio elipse

ellipse.setHeight(20)
ellipse.setWidth(45)
ellipse.setcolor1('yellow') # Probamos los set que ayudan a que cambien los valores del elipse instanciado
ellipse.setcolor2('green') # Para luego graficar y demostrar que realmente cambiaron los datos exitosamente
ellipse.drawEllipse()
```

# Conclusiones

- Las clases permiten agrupar datos (atributos) y comportamientos (métodos) relacionados en una sola estructura, representando objetos como el elipse de forma organizada.
- El constructor (__init__) inicializa los atributos de cada instancia al momento de crearla, permitiendo definir valores por defecto como color1='r' y color2='b'.
- Los métodos set permiten modificar los atributos de un objeto ya creado sin necesidad de instanciarlo nuevamente, lo que resulta más eficiente cuando solo se quiere actualizar parte de sus datos.
- El uso de self dentro de los métodos es lo que permite acceder y modificar los atributos propios de cada instancia, diferenciándolos de variables locales o de otras instancias de la misma clase.
