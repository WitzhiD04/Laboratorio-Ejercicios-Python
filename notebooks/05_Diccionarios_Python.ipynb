```python
soundtrack_dic = {"The Bodyguard":"1992", "Saturday Night Fever":"1977"}
print(soundtrack_dic )

print(soundtrack_dic.keys()) # Llaves del diccionario

print(soundtrack_dic.values()) # Valores en general del diccionario

album_sales_dict = {"Back in Black":50, "The Bodyguard":50, "Thriller":65}
print(album_sales_dict)

print(album_sales_dict["Thriller"]) # Thriller es la llave, se imprimira el num de ventas

print(album_sales_dict.keys())

print(album_sales_dict.values())
```

# Conclusiones

- Los diccionarios almacenan información en pares clave-valor, permitiendo acceder a un dato directamente a través de su llave en lugar de un índice numérico.
- El método keys() devuelve todas las llaves de un diccionario, mientras que values() devuelve todos los valores asociados, ambos como objetos iterables.
- A diferencia de listas y tuplas, el orden de los elementos no es relevante para acceder a la información, ya que la búsqueda se hace por llave.
- Esta estructura resulta muy útil para representar datos con una relación clara entre un identificador (como el nombre de un álbum) y su información asociada (como sus ventas o año de lanzamiento).
