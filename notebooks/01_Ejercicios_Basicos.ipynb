## Actividad 1: Operaciones con Cadenas

```python
A = "1"
print(A)

B = "2"
print(B)

C = A + B
print(C)

D = "ABCDEFG"
print(D[0:3]) # Solo con escoger los 3 primeros en la cadena ya es suficiente para el slicing

E = 'clocrkr1e1c1t'
print(E[::2]) # El enunciado pide imprimir cada 2 elemento, por lo que hacemos un stride con el numero 2

print("\\") # Doble barra invertida para imprimirla, si no la coge para otra funcionalidad, como hacer \n

F = "You are wrong"

F = F.upper() # Upper pasa toda la cadena a mayúsculas

print(F)

G = "Mary had a little lamb Little lamb, little lamb Mary had a little lamb \
Its fleece was white as snow And everywhere that Mary went Mary went, Mary went \
Everywhere that Mary went The lamb was sure to go"

print(G.find('snow')) #Find imprime cuando encuentre el primer snow

G = G.replace("Mary", "Bob") # Con esto aseguramos que todo Mary sea cambiado por Bob

print(G)
```

## Conclusiones

- El *slicing* y el *stride* permiten extraer y transformar subcadenas de forma muy compacta, sin necesidad de bucles explícitos.
- Métodos como `upper()`, `find()` y `replace()` facilitan tareas comunes de limpieza y búsqueda de texto, algo esencial en el preprocesamiento de datos.
- Los strings en Python son inmutables: operaciones como `replace()` devuelven una cadena nueva en lugar de modificar la original.
