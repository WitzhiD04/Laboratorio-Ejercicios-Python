## Actividad: Definición de Funciones

```python
import random

def div(a,b):
    print(a,"dividido", b)
    return a/b

print(div(random.randint(0,1000),random.randint(1,1000))) # Usamos nuevamente random para variedad de resultados
#asi se prueba que se hizo bien 
# b empieza en 1 para evitar divisiones entre 0
print("\n")

def con(a, b):
    return(a + b) #si se puede para ambos cadenas y enteros ya que puede tanto sumar numero como concatenar cadenas

print(con(2,2))
print(con("a","b"))

#Tambien se puede con listas y tuplas, lo que hace simplemente es juntar ambas en una sola

lista1 = [1,2,3,4]
lista2 = [5,6,7,8]
tupla1 = (6,2,10,373)
tupla2 = (34,67,60,47)
print(con(lista1,lista2))
print(con(tupla1,tupla2)) 
```


# Conclusiones

- Las funciones permiten encapsular un bloque de código reutilizable, evitando repetir la misma lógica varias veces en el programa.
- El operador + dentro de una función se comporta de forma distinta según el tipo de dato recibido: suma en números, concatena en cadenas y une elementos en listas y tuplas.
- Este comportamiento demuestra que Python no exige declarar el tipo de los parámetros, por lo que una misma función puede operar sobre distintos tipos de datos siempre que soporten la operación usada dentro de ella.
- El uso de return permite que el resultado de una función pueda almacenarse o utilizarse en otras partes del código, a diferencia de simplemente imprimir el valor dentro de la función.
