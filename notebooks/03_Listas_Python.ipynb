## Actividad: Manejo de Listas

```python
a_list = [1, "hello", [1,2,3], True] # Aprovechamos que las listas permiten amplia variedad en el tipo de datos que permiten
print(a_list)

print(a_list[1])

print(a_list[1:4])

A = [1, 'a']
B = [2, 1, 'd']

C = A + B # Funciona igual que en cadenas, solo con sumar concatena las listas y las junta en una nueva variable
print(C)
```

# Conclusiones

- Las listas permiten almacenar elementos de distintos tipos, incluso otras listas, lo que las hace más flexibles que las tuplas.
- El indexado y el *slicing* funcionan de forma similar a las cadenas y tuplas, reforzando la consistencia de estas operaciones en Python.
- El operador `+` concatena listas creando una nueva, sin alterar las listas originales, a diferencia de métodos como `append()` que sí las modifican.
- Estas propiedades hacen de las listas una estructura versátil para el almacenamiento y manipulación de datos en tareas de procesamiento posteriores.
