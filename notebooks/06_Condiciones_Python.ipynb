## Actividad: Estructuras Condicionales

```python
import random

disco = 8.5

if(disco > 8):
    print("This album is Amazing!")

disco = random.randint(0,10) # se hizo aleatorio para probar lso 2 casos sin editar el código, por lo que nunca será el mismo
#numero y podrá decir si el album es ok o increible

if(disco > 8):
    print("This album is Amazing!")
else:
    print("this album is ok")

disco2 = random.randint(1960,2000) # Lo mismo, en especial este caso ya que varia mucho según el número que saque
print("año del disco ", disco2)

if disco2 < 1980:
    print("El disco salió previo a 1980")
elif disco2 == 1991 or disco2 == 1993:
    print("El disco salió en 1991 o 1993")
```

# Conclusiones

- La estructura if permite ejecutar un bloque de código únicamente cuando se cumple una condición, mientras que else define una alternativa cuando esta no se cumple.
- elif permite evaluar múltiples condiciones de forma secuencial, ejecutando solo el primer bloque cuyo condicional resulte verdadero.
- El uso de valores aleatorios (random.randint) permitió comprobar el comportamiento del código ante distintos escenarios sin necesidad de modificarlo manualmente cada vez.
- Los operadores lógicos como or permiten combinar varias condiciones dentro de un mismo elif, ampliando la flexibilidad para evaluar casos específicos.
