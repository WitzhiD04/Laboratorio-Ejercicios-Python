```python
genres_tuple = ("pop", "rock", "soul", "hard rock", "soft rock","R&B", "progressive rock", "disco") 

print(len(genres_tuple)) #Len ayuda a conocer las longitudes de todo tipo de lista, tupla o colección

print(genres_tuple[3])

print(genres_tuple[3:6])

print(genres_tuple[0:2])

print(genres_tuple.index("disco")) # Se busca en que indice esta disco con index,en caso de existir en la tupla

C_tuple=(-5, 1, -3)

C_tuple = sorted(C_tuple) # Sorted ordena de menor a mayor, pero crea una lista en vez de tupla
print(C_tuple)
```

# Conclusiones
- Las tuplas son inmutables, por lo que operaciones como `sorted()` no las alteran, sino que devuelven una nueva colección (en este caso, una lista).
- El indexado y el *slicing* funcionan igual que en las cadenas, lo que facilita reutilizar la lógica aprendida previamente con strings.
- Métodos como `index()` y `len()` permiten explorar el contenido de una tupla sin necesidad de recorrerla manualmente.
- La inmutabilidad hace de las tuplas una estructura adecuada
