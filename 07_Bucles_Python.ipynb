```python
for i in range(-5,6): 
    print(i)


Genres=[ 'rock', 'R&B', 'Soundtrack', 'R&B', 'soul', 'pop']

for genre in Genres: # Usamos el for apropiado de listas por conveniencia
    print(genre)

squares=['red', 'yellow', 'green', 'purple', 'blue']

print("\n")
for square in squares:
    print(square)

print("\n")

PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]

ratingActual = 7
i = 0
while ratingActual > 6: # Creamos la variable mayor a 6 para entrar al bucle. Lo que haya menor después de un num menor a 6 no se imprimira, pues saldra del bucle al cumplir condición
    ratingActual = PlayListRatings[i]
    print(ratingActual)
    i=i+1

squares = ['orange', 'orange', 'purple', 'blue', 'orange']
new_squares = []
j = 0
print("\n")
while squares[j] == "orange": # Igual que antes, apenas encuentre un dato que no es orange, saldrá del while por cumplir condición a pesar de que hayan mas orange dentro de la lista
    new_squares.append(squares[j]) # Entraran todos los orange seguidos que hayan (en caso de que el 1er dato en la lista sea orange)
    j=j+1
print(new_squares)
```

# Conclusiones

- El bucle for permite recorrer secuencias como rangos y listas de forma directa, ejecutando una acción por cada elemento sin necesidad de controlar manualmente un índice.
- El bucle while se ejecuta mientras una condición sea verdadera, por lo que resulta útil cuando no se conoce de antemano cuántas iteraciones serán necesarias.
- En un while, el ciclo se detiene apenas la condición deja de cumplirse, aunque existan más elementos en la lista que también la satisfagan más adelante (como se vio con PlayListRatings y los elementos "orange").
- Elegir entre for y while depende de si se necesita recorrer una colección completa (for) o iterar hasta que se cumpla o deje de cumplirse una condición específica (while).
