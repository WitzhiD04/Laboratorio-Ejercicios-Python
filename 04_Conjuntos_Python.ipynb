```python
genres = set(['rap','house','electronic music', 'rap']) # set ayuda a convertir listas en conjuntos
#por lo que quitará elementos repetidos como en este caso rap
print(genres)

A = [1, 2, 2, 1]
B = set([1, 2, 2, 1])

print(sum(A) == sum(B)) #Falso ya que el set quita números repetidos, lo que hace que cambia la suma al quitar un 2 

album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])

album_set3 = album_set1 | album_set2 # | representa unión, por lo que juntará todas en un solo conjunto
print(album_set3)

print(album_set1.issubset(album_set3)) # devolverá true, ya que todo album de set1 está en set3. Con is subset se confirma
```

# Conclusiones

- Los conjuntos (sets) eliminan automáticamente los elementos duplicados al crearse, lo que los diferencia de las listas y tuplas.
- Esta eliminación de duplicados puede alterar resultados de operaciones aritméticas, como se vio al comparar sum(A) == sum(B) entre una lista y su versión como set.
- El operador | permite unir dos conjuntos en uno solo, combinando sus elementos sin repetir los que ya existen en ambos.
- El método issubset() permite verificar si todos los elementos de un conjunto están contenidos en otro, siendo útil para comparar relaciones entre colecciones de datos.
